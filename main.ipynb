```python
headers = ['verifier', 'TP', 'FP', 'TN', 'FN', 'soundness', 'completeness']
# create an empty dataframe with headers

results_table = pd.DataFrame(columns=headers)
subdirs = [name for name in os.listdir(verification_results_dir) if os.path.isdir(os.path.join(verification_results_dir, name))]
subdirs.sort()
print(subdirs)
# add rows to the dataframe
for i in subdirs:
    results_table.loc[len(results_table)]=[i, 0, 0, 0, 0, 0, 0]

# print(subdirs)
print(results_table.to_string()) 
```

```python
detail_headers = ['verifier', 'benchmark', 'TP', 'FP', 'unknownFP','TN','FN', 'unknownFN']
benchmarks = ['acasxu','collins_rul_cnn', 'mnist_fc', 'rl_benchmarks', 'tllverifybench', 'traffic_signs_recognition','vggnet16']

detail_results_table = pd.DataFrame(columns=detail_headers)
for i in subdirs:
    for j in benchmarks:
        detail_results_table.loc[len(detail_results_table)]=[i, j, 0, 0, 0, 0, 0, 0]
print(detail_results_table.to_string())
```

```python
for index, row in df.iterrows():
    row_keys = row.iloc[0:3].values
    for col in df.columns:
        if col == 'benchmark':
            benchmark = row[col]
            if benchmark not in detail_results_table['benchmark'].values:
                raise ValueError(f"Benchmark: {benchmark} not in detail results_table!")
        if col.endswith('result'):
            verifier = col.split('_')[0]

            verification_result = row[col]

            verdict = row[verifier+'_true']

            if verifier not in detail_results_table['verifier'].values:
                raise ValueError(f"Verifier: {verifier} not in results_table!")
            
            idx = detail_results_table.index[(detail_results_table['verifier'] == verifier) & (detail_results_table['benchmark'] == benchmark)].tolist()[0]

            if verification_result == 'unsat' and verdict == 'correct':
                detail_results_table.at[idx, 'TP'] += 1

            elif verification_result == 'unsat' and verdict == 'wrong':
                detail_results_table.at[idx, 'FP'] += 1

            elif verification_result == 'sat' and verdict == 'correct':
                detail_results_table.at[idx, 'TN'] += 1

            elif verification_result == 'sat' and verdict == 'wrong':
                detail_results_table.at[idx, 'FN'] += 1

            elif verification_result == 'unknown' or pd.isna(verification_result) or verification_result=='':
                matching_row = ground_truth_df[(ground_truth_df.iloc[:,:3] == row_keys).all(axis=1)]
                ground_truth = matching_row.iloc[0,3]
                if ground_truth == 'unsat':
                    detail_results_table.at[idx, 'unknownFN'] += 1
                elif ground_truth == 'sat':
                    detail_results_table.at[idx, 'unknownFP'] += 1
                else:
                    raise ValueError(f"Ground truth: {ground_truth} not in ['unsat', 'sat']")
            else:
                raise ValueError(f"Verification result: {verification_result} not in ['unsat', 'sat', 'unknown']")
            
print(detail_results_table.to_string())
detail_results_table.to_csv('/home/feng/Documents/cge/results/detail_results_table.csv', index=False)
        
```

```python
import pandas as pd
import onnx
df = pd.read_csv('./results/cleaned_all_results.csv')
def get_num_params(model_path):
    
    model = onnx.load(model_path)
    num_params = 0
    # print(model.graph.initializer)
    
    for tensor in model.graph.initializer:
        if len(tensor.dims) == 0:
            num_params += 1
        else:
            num_params += tensor.dims[0]
    return num_params
df['num_params'] = df['model_path'].apply(get_num_params)
```
